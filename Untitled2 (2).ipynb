```python
# تقسيم البيانات لتدريب واختبار
dataset = Dataset.from_pandas(data[["verse", "explanation"]])
dataset = dataset.train_test_split(test_size=0.1, seed=42)

# تحميل الموديل العربي الجاهز
model_name = "UBC-NLP/AraT5-base"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

max_input_length = 64
max_target_length = 160

def preprocess(examples):
    inputs = ["اشرح هذا البيت الشعري: " + v for v in examples["verse"]]
    model_inputs = tokenizer(inputs, max_length=max_input_length, truncation=True, padding="max_length")
    labels = tokenizer(text_target=examples["explanation"], max_length=max_target_length, truncation=True, padding="max_length")
    model_inputs["labels"] = labels["input_ids"]
    return model_inputs

tokenized_dataset = dataset.map(preprocess, batched=True, remove_columns=["verse", "explanation"])
```

```python
data_collator = DataCollatorForSeq2Seq(tokenizer=tokenizer, model=model)

training_args = Seq2SeqTrainingArguments(
    output_dir="./poetry_model_checkpoints",
    eval_strategy="epoch",
    save_strategy="epoch",
    learning_rate=3e-4,
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,
    num_train_epochs=28,
    weight_decay=0.01,
    predict_with_generate=True,
    logging_steps=10,
    save_total_limit=1,
    report_to="none",
)

trainer = Seq2SeqTrainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_dataset["train"],
    eval_dataset=tokenized_dataset["test"],
    processing_class=tokenizer,
    data_collator=data_collator,
)

trainer.train()
```

```python
save_path = "./my_poetry_explainer_model"
trainer.save_model(save_path)
tokenizer.save_pretrained(save_path)
print("تم حفظ الموديل ")

shutil.make_archive("my_poetry_explainer_model", "zip", save_path)
files.download("my_poetry_explainer_model.zip")
```

```python
#تقيم الموديل قبل وبعد التدريب
import torch # Ensure torch is imported for .to(model.device)
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM # Import necessary classes

def evaluate_explanation(generated_text, reference_text):
    """يحسب نسبة تطابق الكلمات بين شرح الموديل والشرح الحقيقي"""
    generated_words = set(generated_text.split())
    reference_words = set(reference_text.split())
    if len(reference_words) == 0:
        return 0
    overlap = len(generated_words & reference_words)
    score = (overlap / len(reference_words)) * 100
    return round(score, 1)

# Function to explain using the model BEFORE training (loads a fresh, untrained model)
def explain_verse_before_training(verse_text):
    # Load a fresh, untrained tokenizer and model for this comparison
    temp_tokenizer = AutoTokenizer.from_pretrained(model_name)
    temp_model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

    input_text = "اشرح هذا البيت الشعري: " + verse_text
    inputs = temp_tokenizer(input_text, return_tensors="pt", max_length=max_input_length, truncation=True)
    inputs = {k: v.to(temp_model.device) for k, v in inputs.items()}
    outputs = temp_model.generate(
        **inputs,
        max_length=max_target_length,
        num_beams=4,
        early_stopping=True,
    )
    explanation = temp_tokenizer.decode(outputs[0], skip_special_tokens=True)
    return explanation

# Function to explain using the model AFTER training (uses the globally trained model)
def explain_verse(verse_text):
    input_text = "اشرح هذا البيت الشعري: " + verse_text
    inputs = tokenizer(input_text, return_tensors="pt", max_length=max_input_length, truncation=True)
    inputs = {k: v.to(model.device) for k, v in inputs.items()}
    outputs = model.generate(
        **inputs,
        max_length=max_target_length,
        num_beams=4,
        no_repeat_ngram_size=2,
        repetition_penalty=1.5,
        early_stopping=True,
    )
    explanation = tokenizer.decode(outputs[0], skip_special_tokens=True)
    return explanation

test_cases = [
    {
        "verse": "وَمُدَجَّجٍ كَرِهَ الكُماةُ نِزالَهُ لا مُمْعِنٍ هَرَبًا وَلا مُستَسلِم",
        "real_explanation": "يصف الشاعر محارباً شجاعاً مدرّعاً بالكامل بالسلاح، على درجة عالية من القوة والبأس، لدرجة أن أشجع الفرسان يكرهون مواجهته. وهو لا يهرب جباناً من ساحة القتال، ولا يستسلم لخصمه مهما اشتدت المعركة، فيجمع بذلك بين الشجاعة والصمود، وهما صفتا الفارس المثالي في الشعر الجاهلي.",
        "poet": "عنترة بن شداد"
    },
]

print("=" * 50)
print("مقارنة النتائج: قبل التدريب مقابل بعد التدريب")
print("=" * 50)

for case in test_cases:
    verse = case["verse"]
    real = case["real_explanation"]

    before = explain_verse_before_training(verse)   #  (قبل التدريب)
    after = explain_verse(verse)                      # (بعد التدريب)

    score_before = evaluate_explanation(before, real)
    score_after = evaluate_explanation(after, real)

    print(f"\nالبيت: {verse}")
    print(f"الشرح قبل التدريب: {before}")
    print(f"نسبة التطابق قبل: {score_before}%")
    print(f"الشرح بعد التدريب: {after}")
    print(f"نسبة التطابق بعد: {score_after}%")
    print("-" * 50)
```
